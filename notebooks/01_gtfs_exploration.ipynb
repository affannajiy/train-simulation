# GTFS exploration

Looks at the cleaned ETS timetable written by `python -m ets_sim.loader`. Run the loader first.

```python
from ets_sim.config import load_config
from ets_sim.loader import load_processed

cfg = load_config()
tt = load_processed(cfg)
tt.stations.head(10)
```

```python
# Trips per origin-destination pair and direction (0 = northbound, 1 = southbound)
tt.trips.groupby(['direction_id', 'origin', 'destination']).size().sort_values(ascending=False)
```

```python
# Everything the loader dropped, truncated or flagged
tt.quality
```

```python
# Station-to-station section lengths on the main line
main = tt.stations[tt.stations['branch_junction'] == ''].sort_values('line_km')
main.assign(section_km=main['line_km'].diff())[['station', 'line_km', 'section_km']]
```
